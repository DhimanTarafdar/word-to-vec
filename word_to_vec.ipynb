# Word2Vec Introduction
### Embedding কী এবং কেন দরকার? → তারপর Word2Vec

---

## 1. Embedding কী?

**Embedding** হলো এমন একটি technique যেখানে আমরা high-dimensional বা discrete data-কে (যেমন words, sentences, images ইত্যাদি) **dense, continuous vector**-এ রূপান্তর করি।

সহজ ভাষায়:
> আমরা কোনো জিনিসকে এমন সংখ্যার list (vector) দিয়ে represent করি, যাতে সেই সংখ্যাগুলোর মধ্যে **semantic meaning** এবং **relationship** ধরা পড়ে।

### উদাহরণ দিয়ে বুঝি (Intuition):

ধরো তোমার কাছে এই শব্দগুলো আছে:
- King
- Queen
- Man
- Woman
- Apple
- Orange

যদি আমরা শুধু one-hot encoding ব্যবহার করি, তাহলে:
- King  → `[1, 0, 0, 0, 0, 0]`
- Queen → `[0, 1, 0, 0, 0, 0]`
- Man   → `[0, 0, 1, 0, 0, 0]`

এখানে কোনো semantic relationship নেই।  
King এবং Queen কতটা কাছাকাছি, বা King - Man + Woman ≈ Queen — এই ধরনের relation one-hot-এ ধরাই যায় না।

**Embedding** এই সমস্যার সমাধান করে।  
Embedding-এ প্রতিটি word-কে একটি dense vector দেওয়া হয়, যেমন:

- King   → [0.2, 0.8, 0.1, ..., 0.4]
- Queen  → [0.3, 0.7, 0.2, ..., 0.5]
- Man    → [0.1, 0.6, 0.0, ..., 0.3]


এখন vector space-এ:
- King এবং Queen কাছাকাছি থাকে
- Man এবং Woman কাছাকাছি থাকে
- এবং বিখ্যাত analogy কাজ করে:  
  **King - Man + Woman ≈ Queen**

এটাই Embedding-এর শক্তি।

---

## 2. Embedding কেন দরকার? (Why do we need Embeddings?)

Neural Network এবং Deep Learning model-গুলো **সংখ্যা** নিয়ে কাজ করে। তারা সরাসরি text বুঝতে পারে না।

### সমস্যাগুলো:

| সমস্যা | ব্যাখ্যা |
|------|--------|
| **Discrete nature** | Words হলো discrete symbols। Model এগুলোকে সরাসরি process করতে পারে না। |
| **No semantic information** | One-hot encoding-এ কোনো meaning বা similarity নেই। |
| **Curse of Dimensionality** | Vocabulary size হাজার হাজার হলে one-hot vector অনেক বড় এবং sparse হয় → computation expensive + generalization খারাপ হয়। |
| **No generalization** | Model দেখেনি এমন word-এর জন্য কিছুই শিখতে পারে না। |

### Embedding কীভাবে সাহায্য করে:

1. **Dense Representation** → কম dimension-এ বেশি information
2. **Semantic Similarity** capture করে
3. **Transfer Learning** সম্ভব হয় (pre-trained embedding ব্যবহার করা যায়)
4. Model-কে **generalization** করতে সাহায্য করে

---

## 3. Word Embedding কী?

যখন আমরা **words**-কে vector-এ রূপান্তর করি, তখন তাকে **Word Embedding** বলা হয়।

Word Embedding-এর লক্ষ্য:
> এমন vector শেখা যাতে semantically similar words vector space-এ কাছাকাছি থাকে এবং linguistic relationships (analogy, gender, tense ইত্যাদি) গাণিতিকভাবে capture হয়।

---

## 4. Word2Vec কী?

**Word2Vec** হলো Word Embedding শেখার সবচেয়ে বিখ্যাত এবং প্রভাবশালী algorithm।

এটি ২০১৩ সালে Google-এর Tomas Mikolov এবং তার team তৈরি করেছিলেন।

### মূল Idea (Intuition):

> **"You shall know a word by the company it keeps."**  
> — J.R. Firth (1957)

মানে: একটি শব্দের অর্থ বোঝা যায় সে যেসব শব্দের সাথে থাকে, সেগুলো দেখে।

Word2Vec এই principle-কে কাজে লাগিয়ে দুটি neural network architecture ব্যবহার করে word embedding শেখে:

1. **CBOW** (Continuous Bag of Words)
2. **Skip-gram**

### Word2Vec কী করে?

- বিশাল পরিমাণ text (corpus) দেখে
- প্রতিটি word-এর জন্য একটি dense vector শেখে
- যেসব words একই context-এ আসে, তাদের vector কাছাকাছি হয়
- ফলে semantic এবং syntactic relationship automatically capture হয়

### কেন Word2Vec এত গুরুত্বপূর্ণ?

| কারণ | ব্যাখ্যা |
|------|--------|
| **Efficiency** | আগের method-গুলোর তুলনায় অনেক দ্রুত এবং scalable |
| **Quality** | খুব ভালো quality-এর embedding তৈরি করে |
| **Analogy Capability** | `King - Man + Woman ≈ Queen` এরকম relation ধরে |
| **Foundation** | পরবর্তী অনেক advanced model (GloVe, FastText, BERT ইত্যাদি)-এর ভিত্তি তৈরি করেছে |

# Word2Vec — Brief Idea & Core Intuition
### Embedding তৈরির পিছনের আসল চিন্তা কী? কীভাবে কাজ করে?

---

## 1. মূল Intuition (সবচেয়ে গুরুত্বপূর্ণ অংশ)

Word2Vec-এর পুরো কাজের ভিত্তি একটাই খুব শক্তিশালী idea:

> **"একটা শব্দের অর্থ বোঝা যায় সে যেসব শব্দের সাথে থাকে, সেগুলো দেখে।"**  
> (Distributional Hypothesis)

মানে:
- যেসব শব্দ **একই ধরনের context**-এ আসে → তাদের অর্থ কাছাকাছি
- যেসব শব্দ **আলাদা context**-এ আসে → তাদের অর্থ আলাদা

এটাই Word2Vec শেখে।

---

## 2. Perfect Example দিয়ে বুঝি

ধরো আমাদের কাছে একটা ছোট corpus আছে:

The cat is sleeping on the sofa

The dog is sleeping on the bed

The cat is running in the garden

The dog is running in the park

The car is parked near the house

The car is moving on the road


এখন Word2Vec প্রতিটি শব্দের চারপাশের শব্দগুলো (context) দেখে।

### Context Window উদাহরণ (window size = 2):

| Target Word | Context Words |
|-------------|---------------|
| **cat**     | the, is, sleeping, on |
| **dog**     | the, is, sleeping, on |
| **cat**     | the, is, running, in |
| **dog**     | the, is, running, in |
| **car**     | the, is, parked, near |
| **car**     | the, is, moving, on |

### লক্ষ্য করো কী হচ্ছে:

- **cat** এবং **dog** প্রায় একই context-এ আসছে → `sleeping`, `running`, `is`, `the`
- **car** সম্পূর্ণ আলাদা context-এ আসছে → `parked`, `moving`, `road`, `house`

Word2Vec এই pattern দেখে বুঝে যায়:

> "cat এবং dog-এর আশেপাশের শব্দগুলো একই রকম → তাই তাদের vector কাছাকাছি হওয়া উচিত"  
> "car-এর context আলাদা → তার vector দূরে থাকা উচিত"

---

## 3. Word থেকে Dense Vector-এ যাওয়ার প্রক্রিয়া (High-level)

Word2Vec সরাসরি meaning শেখে না। সে শেখে **prediction** করতে।

### দুইভাবে কাজ করতে পারে:

**A. Skip-gram:**
> "আমি target word দেখে তার আশেপাশের context words predict করব"

**B. CBOW:**
> "আমি context words দেখে target word-টা predict করব"

যেভাবেই হোক, model-কে ভালো করে predict করতে হলে তাকে এমন vector শিখতে হয় যেখানে:

- একই context-এর words → কাছাকাছি vector পায়
- আলাদা context-এর words → দূরের vector পায়

### Visualize করো:

প্রথমে প্রতিটি word-কে randomly একটা vector দেওয়া হয় (যেমন 100 বা 300 dimension-এর)।

তারপর training চলতে থাকে।  
প্রতিবার যখন model ভুল predict করে, vectorগুলোকে সামান্য করে adjust করা হয়।

অনেকবার এভাবে adjust করার পর:
cat  →  [0.21,  0.87, -0.14,  0.55, ...,  0.33]

dog  →  [0.19,  0.84, -0.11,  0.52, ...,  0.31]   ← cat-এর খুব কাছাকাছি

car  →  [0.78, -0.22,  0.61, -0.45, ..., -0.17]   ← অনেক দূরে


এখন যদি আমরা cosine similarity মাপি:

- similarity(cat, dog) → খুব বেশি (যেমন 0.92)
- similarity(cat, car) → কম (যেমন 0.18)
- similarity(dog, car) → কম

---

## 4. কীভাবে বোঝে "cat/dog" এক type, "car" আরেক type?

Word2Vec কখনো সরাসরি বলে না যে "এগুলো animal" বা "এটা vehicle"।

সে শুধু **statistical pattern** থেকে শেখে:

| Pattern | Word2Vec কী শেখে |
|-------|------------------|
| cat এবং dog দুটোই "sleeping", "running", "eating" এর সাথে আসে | এদের vector কাছাকাছি → একই group |
| car "parked", "driving", "engine", "road" এর সাথে আসে | আলাদা group |
| "cat" এবং "tiger" দুটোই "roaring", "hunting" এর সাথে আসতে পারে | আরও কাছাকাছি হতে পারে |

ফলে vector space-এ automatically cluster তৈরি হয়:

- Animal cluster → cat, dog, tiger, lion...
- Vehicle cluster → car, bus, truck, bike...
- Food cluster → rice, bread, apple...

এটাই **unsupervised** ভাবে semantic meaning capture করার শক্তি।

---

## সংক্ষেপে পুরো Idea:

1. Word2Vec শব্দের **অর্থ** সরাসরি শেখে না।
2. সে শেখে শব্দগুলো **কোন context**-এ আসে।
3. একই context → কাছাকাছি vector
4. আলাদা context → দূরের vector
5. ফলে dense vector-এর মধ্যে semantic relationship চলে আসে।

# Dense Vector-এর Dimension এবং Value Update হওয়ার প্রক্রিয়া
### ছোট Example দিয়ে পুরোটা পরিষ্কারভাবে

---

## 1. Dense Vector-এর Dimension (Shape) কীভাবে হয়?

**Dimension আমরা নিজেরাই ঠিক করি।** এটা একটা **hyperparameter**।

### মানে কী?

যখন আমরা Word2Vec model তৈরি করি, তখন আমরা বলে দিই:

> "প্রতিটি word-কে আমি কয়টা সংখ্যা দিয়ে represent করতে চাই?"

এই সংখ্যাটাকেই বলে **Embedding Dimension** বা **Vector Size**।

| Common Choice | ব্যবহার |
|---------------|--------|
| 50 – 100      | ছোট dataset / দ্রুত experiment |
| 200 – 300     | সবচেয়ে বেশি ব্যবহৃত (Google News Word2Vec-ও 300 dim) |
| 500+          | খুব বড় dataset, বেশি capacity লাগলে |

### উদাহরণ:

ধরো আমাদের vocabulary-তে ৫টা word আছে:

["cat", "dog", "car", "run", "sleep"]

textআমরা যদি **Embedding Dimension = 3** রাখি, তাহলে প্রতিটি word-এর vector
হবে ৩টা সংখ্যার:

cat   → [ ?, ?, ? ]

dog   → [ ?, ?, ? ]

car   → [ ?, ?, ? ]

run   → [ ?, ?, ? ]

sleep → [ ?, ?, ? ]


**মনে রাখো:**  
Dimension যত বাড়াবে, model তত বেশি complex relationship ধরতে পারবে, কিন্তু:
- বেশি data লাগবে
- Training সময় বাড়বে
- Overfitting-এর ঝুঁকি বাড়বে

---

## 2. Valueগুলো শুরুতে কেমন থাকে?

Training শুরুর আগে সব vector **random** সংখ্যা দিয়ে initialize করা হয়।

### ছোট Example (Dimension = 2 রাখছি যাতে সহজে বোঝা যায়):

Initial Random Vectors (2D):

cat   → [ 0.10,  0.40]

dog   → [ 0.30, -0.20]

car   → [-0.50,  0.60]

run   → [ 0.70,  0.10]

sleep → [-0.20, -0.40]


এখন এদের মধ্যে কোনো meaningful relation নেই। সব random।

---

## 3. Valueগুলো কীভাবে Update হয়? (সবচেয়ে গুরুত্বপূর্ণ অংশ)

Word2Vec একটা **prediction** খেলা খেলে।

### ধরো আমরা Skip-gram ব্যবহার করছি:

**Task:** Target word দেখে তার পাশের word predict করা।

Corpus থেকে একটা উদাহরণ নিলাম:
> "cat is sleeping"

Target = **cat**  
Context = **sleeping**

Model-কে বলা হয়:
> "cat এর vector দিয়ে বলো, পাশের wordটা 'sleeping' হওয়ার সম্ভাবনা কত?"

### ধাপে ধাপে কী হয়:

**Step 1: Prediction**
- cat-এর বর্তমান vector নেয় → `[0.10, 0.40]`
- এই vector দিয়ে সব word-এর সাথে score হিসাব করে (dot product)
- Softmax বা Negative Sampling দিয়ে probability বের করে

**Step 2: Loss হিসাব**
- সঠিক word ছিল "sleeping"
- Model যদি ভুল predict করে (অর্থাৎ "sleeping"-এর probability কম দেয়), তাহলে **Loss** বেশি হয়

**Step 3: Gradient Descent দিয়ে Update**
- Loss কমানোর জন্য vectorগুলোকে সামান্য করে পরিবর্তন করা হয়
- **cat** এবং **sleeping** এর vectorকে একে অপরের কাছাকাছি নিয়ে আসা হয়
- অন্য unrelated word-এর vectorকে একটু দূরে ঠেলে দেওয়া হয়

এই process হাজার হাজার বার চলে (পুরো corpus বহুবার দেখে)।

---

## 4. ছোট Numerical Example (খুবই Simplified)

ধরো শুধু ২টা word নিয়ে দেখি: **cat** এবং **sleep**


Initial:

cat   → [0.10, 0.40]

sleep → [-0.20, -0.40]

textএদের Dot Product = (0.10 × -0.20) + (0.40 × -0.40) = -0.02 - 0.16 = **-0.18** (খুব কম → খারাপ)


Training-এর পর (অনেক update-এর পর) ধরো হয়ে গেল:

cat   → [0.35, 0.60]

sleep → [0.40, 0.55]

textএখন Dot Product = (0.35×0.40) + (0.60×0.55) = 0.14 + 0.33 = **0.47** (বেশি হয়েছে → ভালো)


মানে model শিখেছে যে "cat" এবং "sleep" একসাথে আসে, তাই তাদের vector কাছাকাছি নিয়ে এসেছে।

# CBOW (Continuous Bag of Words)
### Word2Vec-এর একটি গুরুত্বপূর্ণ Architecture

---

## 1. CBOW কী?

**CBOW** মানে **Continuous Bag of Words**।

এটা Word2Vec-এর দুইটা মূল architecture-এর একটি (অন্যটা Skip-gram)।

### মূল Idea:

> **আশেপাশের context wordsগুলো দেখে মাঝখানের target word-টা predict করা।**

সহজ ভাষায়:
- তুমি যদি বলো: “The ___ is sleeping on the sofa”
- CBOW context দেখে বলবে → সম্ভাব্য word হলো **cat** বা **dog**

এখানে “Bag of Words” বলা হয় কারণ context words-এর **ক্রম (order)** কে তেমন গুরুত্ব দেয় না। শুধু কোন কোন word আছে সেটা দেখে।

---

## 2. CBOW কেন ব্যবহার করা হয়?

| সুবিধা | ব্যাখ্যা |
|-------|--------|
| **দ্রুত Training** | Skip-gram-এর তুলনায় অনেক দ্রুত শেখে |
| **Frequent words-এ ভালো** | যেসব word বারবার আসে (the, is, cat, dog ইত্যাদি), সেগুলোর ভালো embedding তৈরি করে |
| **Smooth learning** | অনেক context word একসাথে ব্যবহার করায় noise কম হয় |
| **কম Data-তেও কাজ করে** | ছোট dataset-এও যথেষ্ট ভালো ফল দেয় |

**অসুবিধা:**
- Rare words-এর embedding তুলনামূলকভাবে দুর্বল হয় (কারণ rare word target হিসেবে কম আসে)

---

## 3. CBOW কীভাবে কাজ করে?

### Architecture (সহজভাবে):

Context Words → Average / Sum → Hidden Layer → Predict Target Word


### ধাপে ধাপে প্রক্রিয়া:

1. একটা window size ঠিক করা হয় (যেমন window = 2)
2. Target word-এর আগে ও পরে ২টা করে word নেওয়া হয় (context)
3. এই context words-এর embedding vectorগুলোকে **average** করা হয়
4. এই average vector দিয়ে target word-টা predict করার চেষ্টা করা হয়
5. ভুল হলে Loss হিসাব করে embeddingগুলো update করা হয়

## 5. কখন / কোন ধরনের কাজে CBOW ব্যবহার করব? (সবচেয়ে গুরুত্বপূর্ণ)

### CBOW ব্যবহার করার সঠিক সময়:

| পরিস্থিতি | কেন CBOW ভালো |
|----------|----------------|
| **Dataset বড় এবং Frequent words বেশি** | দ্রুত এবং ভালো quality পাওয়া যায় |
| **Training সময় কম রাখতে চাও** | Skip-gram-এর চেয়ে অনেক দ্রুত |
| **General purpose embedding** চাও | বেশিরভাগ common word-এর জন্য যথেষ্ট ভালো |
| **Resource কম** (GPU/CPU কম) | কম computation লাগে |
| **Downstream task-এ frequent words গুরুত্বপূর্ণ** | যেমন Sentiment Analysis, Text Classification |

### কোন কাজে / Research-এ বেশি ব্যবহার হয়:

- **Text Classification** (খবরের ধরন বোঝা, spam detection ইত্যাদি)
- **Sentiment Analysis**
- **Named Entity Recognition (NER)**-এর জন্য pre-trained embedding হিসেবে
- **Information Retrieval**
- যখন তুমি **দ্রুত baseline** বানাতে চাও
- বড় corpus (যেমন Wikipedia, news articles) দিয়ে embedding train করতে

### কখন CBOW ব্যবহার করবে না:

- যদি তোমার কাছে **অনেক rare words** থাকে এবং সেগুলো গুরুত্বপূর্ণ হয় → তখন **Skip-gram** ভালো
- যদি খুব ছোট dataset হয় এবং rare words-ও শিখতে চাও → Skip-gram ভালো

> **CBOW = Context দেখে Target predict করে**  
> এটা দ্রুত, frequent words-এ শক্তিশালী, এবং বেশিরভাগ সাধারণ NLP কাজে ভালো কাজ করে।

# Skip-gram
### Word2Vec-এর আরেকটি গুরুত্বপূর্ণ Architecture

---

## 1. Skip-gram কী?

**Skip-gram** হলো Word2Vec-এর দ্বিতীয় মূল architecture।

### মূল Idea:

> **একটা target word দেখে তার আশেপাশের context wordsগুলো predict করা।**

CBOW-এর ঠিক উল্টো।

সহজ ভাষায়:
- তুমি যদি শুধু **"sleeping"** শব্দটা দাও
- Skip-gram চেষ্টা করবে বলতে → এর আশেপাশে "cat", "is", "on", "sofa" থাকতে পারে

একে **Skip-gram** বলা হয় কারণ এটা মাঝখানের word দিয়ে চারপাশের wordগুলো predict করার সময় কিছু word “skip” করেও context ধরতে পারে।

---

## 2. Skip-gram কেন ব্যবহার করা হয়?

| সুবিধা | ব্যাখ্যা |
|-------|--------|
| **Rare words-এ অসাধারণ** | কম আসা words-এরও ভালো embedding তৈরি করে |
| **বেশি Quality** | সাধারণত CBOW-এর চেয়ে ভালো semantic relationship ধরে |
| **Analogy-তে শক্তিশালী** | `King - Man + Woman ≈ Queen` এরকম কাজ বেশি ভালো করে |
| **ছোট Dataset-এও ভালো** | কম ডাটা থাকলেও যথেষ্ট useful embedding দেয় |

**অসুবিধা:**
- Training সময় বেশি লাগে (CBOW-এর তুলনায় ধীর)
- Frequent words-এর জন্য অতিরিক্ত computational cost হয়

---

## 3. Skip-gram কীভাবে কাজ করে?

### Architecture (সহজভাবে):

Target Word → Hidden Layer → Predict Context Words


### ধাপে ধাপে প্রক্রিয়া:

1. Window size ঠিক করা হয় (যেমন window = 2)
2. একটা target word নেওয়া হয়
3. সেই target word দিয়ে তার আগে-পরের context wordsগুলো predict করার চেষ্টা করা হয়
4. প্রতিটি context word-এর জন্য আলাদা করে prediction হয়
5. ভুল হলে Loss হিসাব করে target এবং context দুটোর embedding-ই update করা হয়

---

## 4. ছোট Example দিয়ে বুঝি

Sentence:
> **"The cat is sleeping on the sofa"**

ধরি window size = 2 এবং target word = **"sleeping"**

তাহলে Skip-gram যা predict করার চেষ্টা করবে:

Input:   sleeping

↓

Predict করবে → cat, is, on, the


অর্থাৎ modelকে বলা হয়:
> “sleeping” শব্দটা দেখে বলো, এর চারপাশে কোন কোন শব্দ থাকতে পারে।

### আরেকটা উদাহরণ:

Target = **"cat"**  
Context = `["The", "is", "sleeping", "on"]`

বারবার এভাবে train হতে থাকলে:
- "cat" যখনই আসে, তার সাথে "sleeping", "running", "eating" ইত্যাদি predict করতে হয়
- "dog"-ও একই ধরনের context predict করতে হয়
- ফলে "cat" এবং "dog"-এর vector কাছাকাছি চলে আসে
- আর "car"-এর সাথে এই contextগুলো না আসায় সে দূরে থাকে

---

## 5. কখন / কোন ধরনের কাজে Skip-gram ব্যবহার করব?

### Skip-gram ব্যবহার করার সঠিক সময়:

| পরিস্থিতি | কেন Skip-gram ভালো |
|----------|---------------------|
| **Rare words গুরুত্বপূর্ণ** | চিকিৎসা, আইন, scientific text ইত্যাদিতে |
| **সবচেয়ে ভালো quality চাও** | Semantic similarity ও analogy বেশি ভালো হয় |
| **Dataset ছোট বা medium** | কম ডাটাতেও ভালো represent শেখে |
| **Downstream task-এ rare words matter করে** | যেমন Medical NLP, Legal NLP, Technical domain |
| **Research কাজ** | Paper-এ সাধারণত Skip-gram বেশি ব্যবহার হয় কারণ quality ভালো |

### কোন কাজে / Research-এ বেশি ব্যবহার হয়:

- **Word Similarity** এবং **Word Analogy** task
- **Medical / Biomedical NLP** (rare medical terms)
- **Legal document analysis**
- **Recommendation System** (product names, rare items)
- **Low-resource language**-এ embedding তৈরি
- যখন তুমি **highest quality embedding** চাও এবং সময়ের ব্যাপারে কম চিন্তা করো

### কখন Skip-gram ব্যবহার করবে না:

- যদি খুব বড় corpus হয় এবং শুধু দ্রুত baseline লাগে → CBOW ভালো
- যদি শুধু frequent words-ই গুরুত্বপূর্ণ হয় এবং সময় বাঁচাতে চাও → CBOW ভালো

---

## CBOW vs Skip-gram (দ্রুত তুলনা)

| বিষয়              | CBOW                          | Skip-gram                     |
|-------------------|-------------------------------|-------------------------------|
| কাজ               | Context → Target              | Target → Context              |
| গতি               | দ্রুত                         | ধীর                           |
| Frequent words    | খুব ভালো                      | ভালো                          |
| Rare words        | দুর্বল                        | খুব ভালো                      |
| Embedding Quality | ভালো                          | সাধারণত আরও ভালো              |
| ব্যবহার           | দ্রুত baseline, বড় ডাটা       | Research, rare words, quality |

---

## সংক্ষেপে মনে রাখার মতো কথা:

> **Skip-gram = Target word দিয়ে Context predict করে**  
> এটা ধীর, কিন্তু rare words-এ শক্তিশালী এবং overall embedding quality বেশি ভালো হয়।  
> Research এবং high-quality কাজের জন্য সাধারণত Skip-gram-কে prefer করা হয়।
