# Word2Vec Introduction
### Embedding কী এবং কেন দরকার? → তারপর Word2Vec

---

## 1. Embedding কী?

**Embedding** হলো এমন একটি technique যেখানে আমরা high-dimensional বা discrete data-কে (যেমন words, sentences, images ইত্যাদি) **dense, continuous vector**-এ রূপান্তর করি।

সহজ ভাষায়:
> আমরা কোনো জিনিসকে এমন সংখ্যার list (vector) দিয়ে represent করি, যাতে সেই সংখ্যাগুলোর মধ্যে **semantic meaning** এবং **relationship** ধরা পড়ে।

### উদাহরণ দিয়ে বুঝি (Intuition):

ধরো তোমার কাছে এই শব্দগুলো আছে:
- King
- Queen
- Man
- Woman
- Apple
- Orange

যদি আমরা শুধু one-hot encoding ব্যবহার করি, তাহলে:
- King  → `[1, 0, 0, 0, 0, 0]`
- Queen → `[0, 1, 0, 0, 0, 0]`
- Man   → `[0, 0, 1, 0, 0, 0]`

এখানে কোনো semantic relationship নেই।  
King এবং Queen কতটা কাছাকাছি, বা King - Man + Woman ≈ Queen — এই ধরনের relation one-hot-এ ধরাই যায় না।

**Embedding** এই সমস্যার সমাধান করে।  
Embedding-এ প্রতিটি word-কে একটি dense vector দেওয়া হয়, যেমন:

- King   → [0.2, 0.8, 0.1, ..., 0.4]
- Queen  → [0.3, 0.7, 0.2, ..., 0.5]
- Man    → [0.1, 0.6, 0.0, ..., 0.3]


এখন vector space-এ:
- King এবং Queen কাছাকাছি থাকে
- Man এবং Woman কাছাকাছি থাকে
- এবং বিখ্যাত analogy কাজ করে:  
  **King - Man + Woman ≈ Queen**

এটাই Embedding-এর শক্তি।

---

## 2. Embedding কেন দরকার? (Why do we need Embeddings?)

Neural Network এবং Deep Learning model-গুলো **সংখ্যা** নিয়ে কাজ করে। তারা সরাসরি text বুঝতে পারে না।

### সমস্যাগুলো:

| সমস্যা | ব্যাখ্যা |
|------|--------|
| **Discrete nature** | Words হলো discrete symbols। Model এগুলোকে সরাসরি process করতে পারে না। |
| **No semantic information** | One-hot encoding-এ কোনো meaning বা similarity নেই। |
| **Curse of Dimensionality** | Vocabulary size হাজার হাজার হলে one-hot vector অনেক বড় এবং sparse হয় → computation expensive + generalization খারাপ হয়। |
| **No generalization** | Model দেখেনি এমন word-এর জন্য কিছুই শিখতে পারে না। |

### Embedding কীভাবে সাহায্য করে:

1. **Dense Representation** → কম dimension-এ বেশি information
2. **Semantic Similarity** capture করে
3. **Transfer Learning** সম্ভব হয় (pre-trained embedding ব্যবহার করা যায়)
4. Model-কে **generalization** করতে সাহায্য করে

---

## 3. Word Embedding কী?

যখন আমরা **words**-কে vector-এ রূপান্তর করি, তখন তাকে **Word Embedding** বলা হয়।

Word Embedding-এর লক্ষ্য:
> এমন vector শেখা যাতে semantically similar words vector space-এ কাছাকাছি থাকে এবং linguistic relationships (analogy, gender, tense ইত্যাদি) গাণিতিকভাবে capture হয়।

---

## 4. Word2Vec কী?

**Word2Vec** হলো Word Embedding শেখার সবচেয়ে বিখ্যাত এবং প্রভাবশালী algorithm।

এটি ২০১৩ সালে Google-এর Tomas Mikolov এবং তার team তৈরি করেছিলেন।

### মূল Idea (Intuition):

> **"You shall know a word by the company it keeps."**  
> — J.R. Firth (1957)

মানে: একটি শব্দের অর্থ বোঝা যায় সে যেসব শব্দের সাথে থাকে, সেগুলো দেখে।

Word2Vec এই principle-কে কাজে লাগিয়ে দুটি neural network architecture ব্যবহার করে word embedding শেখে:

1. **CBOW** (Continuous Bag of Words)
2. **Skip-gram**

### Word2Vec কী করে?

- বিশাল পরিমাণ text (corpus) দেখে
- প্রতিটি word-এর জন্য একটি dense vector শেখে
- যেসব words একই context-এ আসে, তাদের vector কাছাকাছি হয়
- ফলে semantic এবং syntactic relationship automatically capture হয়

### কেন Word2Vec এত গুরুত্বপূর্ণ?

| কারণ | ব্যাখ্যা |
|------|--------|
| **Efficiency** | আগের method-গুলোর তুলনায় অনেক দ্রুত এবং scalable |
| **Quality** | খুব ভালো quality-এর embedding তৈরি করে |
| **Analogy Capability** | `King - Man + Woman ≈ Queen` এরকম relation ধরে |
| **Foundation** | পরবর্তী অনেক advanced model (GloVe, FastText, BERT ইত্যাদি)-এর ভিত্তি তৈরি করেছে |

## সংক্ষেপে Flow:
Text (Words)

↓

↓

Embedding Layer / Word Embedding

↓

Dense Vector (ভালো)

↓

Word2Vec → সবচেয়ে জনপ্রিয় Word Embedding technique

# Word2Vec — Brief Idea & Core Intuition
### Embedding তৈরির পিছনের আসল চিন্তা কী? কীভাবে কাজ করে?

---

## 1. মূল Intuition (সবচেয়ে গুরুত্বপূর্ণ অংশ)

Word2Vec-এর পুরো কাজের ভিত্তি একটাই খুব শক্তিশালী idea:

> **"একটা শব্দের অর্থ বোঝা যায় সে যেসব শব্দের সাথে থাকে, সেগুলো দেখে।"**  
> (Distributional Hypothesis)

মানে:
- যেসব শব্দ **একই ধরনের context**-এ আসে → তাদের অর্থ কাছাকাছি
- যেসব শব্দ **আলাদা context**-এ আসে → তাদের অর্থ আলাদা

এটাই Word2Vec শেখে।

---

## 2. Perfect Example দিয়ে বুঝি

ধরো আমাদের কাছে একটা ছোট corpus আছে:

The cat is sleeping on the sofa

The dog is sleeping on the bed

The cat is running in the garden

The dog is running in the park

The car is parked near the house

The car is moving on the road


এখন Word2Vec প্রতিটি শব্দের চারপাশের শব্দগুলো (context) দেখে।

### Context Window উদাহরণ (window size = 2):

| Target Word | Context Words |
|-------------|---------------|
| **cat**     | the, is, sleeping, on |
| **dog**     | the, is, sleeping, on |
| **cat**     | the, is, running, in |
| **dog**     | the, is, running, in |
| **car**     | the, is, parked, near |
| **car**     | the, is, moving, on |

### লক্ষ্য করো কী হচ্ছে:

- **cat** এবং **dog** প্রায় একই context-এ আসছে → `sleeping`, `running`, `is`, `the`
- **car** সম্পূর্ণ আলাদা context-এ আসছে → `parked`, `moving`, `road`, `house`

Word2Vec এই pattern দেখে বুঝে যায়:

> "cat এবং dog-এর আশেপাশের শব্দগুলো একই রকম → তাই তাদের vector কাছাকাছি হওয়া উচিত"  
> "car-এর context আলাদা → তার vector দূরে থাকা উচিত"

---

## 3. Word থেকে Dense Vector-এ যাওয়ার প্রক্রিয়া (High-level)

Word2Vec সরাসরি meaning শেখে না। সে শেখে **prediction** করতে।

### দুইভাবে কাজ করতে পারে:

**A. Skip-gram:**
> "আমি target word দেখে তার আশেপাশের context words predict করব"

**B. CBOW:**
> "আমি context words দেখে target word-টা predict করব"

যেভাবেই হোক, model-কে ভালো করে predict করতে হলে তাকে এমন vector শিখতে হয় যেখানে:

- একই context-এর words → কাছাকাছি vector পায়
- আলাদা context-এর words → দূরের vector পায়

### Visualize করো:

প্রথমে প্রতিটি word-কে randomly একটা vector দেওয়া হয় (যেমন 100 বা 300 dimension-এর)।

তারপর training চলতে থাকে।  
প্রতিবার যখন model ভুল predict করে, vectorগুলোকে সামান্য করে adjust করা হয়।

অনেকবার এভাবে adjust করার পর:
cat  →  [0.21,  0.87, -0.14,  0.55, ...,  0.33]

dog  →  [0.19,  0.84, -0.11,  0.52, ...,  0.31]   ← cat-এর খুব কাছাকাছি

car  →  [0.78, -0.22,  0.61, -0.45, ..., -0.17]   ← অনেক দূরে


এখন যদি আমরা cosine similarity মাপি:

- similarity(cat, dog) → খুব বেশি (যেমন 0.92)
- similarity(cat, car) → কম (যেমন 0.18)
- similarity(dog, car) → কম

## 4. কীভাবে বোঝে "cat/dog" এক type, "car" আরেক type?

Word2Vec কখনো সরাসরি বলে না যে "এগুলো animal" বা "এটা vehicle"।

সে শুধু **statistical pattern** থেকে শেখে:

| Pattern | Word2Vec কী শেখে |
|-------|------------------|
| cat এবং dog দুটোই "sleeping", "running", "eating" এর সাথে আসে | এদের vector কাছাকাছি → একই group |
| car "parked", "driving", "engine", "road" এর সাথে আসে | আলাদা group |
| "cat" এবং "tiger" দুটোই "roaring", "hunting" এর সাথে আসতে পারে | আরও কাছাকাছি হতে পারে |

ফলে vector space-এ automatically cluster তৈরি হয়:

- Animal cluster → cat, dog, tiger, lion...
- Vehicle cluster → car, bus, truck, bike...
- Food cluster → rice, bread, apple...

এটাই **unsupervised** ভাবে semantic meaning capture করার শক্তি।

## সংক্ষেপে পুরো Idea:

1. Word2Vec শব্দের **অর্থ** সরাসরি শেখে না।
2. সে শেখে শব্দগুলো **কোন context**-এ আসে।
3. একই context → কাছাকাছি vector
4. আলাদা context → দূরের vector
5. ফলে dense vector-এর মধ্যে semantic relationship চলে আসে।
