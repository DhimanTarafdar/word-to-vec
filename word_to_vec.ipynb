## 1. Embedding কী?

**Embedding** হলো এমন একটি technique যেখানে আমরা high-dimensional বা discrete data-কে (যেমন words, sentences, images ইত্যাদি) **dense, continuous vector**-এ রূপান্তর করি।

সহজ ভাষায়:
> আমরা কোনো জিনিসকে এমন সংখ্যার list (vector) দিয়ে represent করি, যাতে সেই সংখ্যাগুলোর মধ্যে **semantic meaning** এবং **relationship** ধরা পড়ে।

### উদাহরণ দিয়ে বুঝি (Intuition):

ধরো তোমার কাছে এই শব্দগুলো আছে:
- King
- Queen
- Man
- Woman
- Apple
- Orange

যদি আমরা শুধু one-hot encoding ব্যবহার করি, তাহলে:
- King  → `[1, 0, 0, 0, 0, 0]`
- Queen → `[0, 1, 0, 0, 0, 0]`
- Man   → `[0, 0, 1, 0, 0, 0]`

এখানে কোনো semantic relationship নেই।  
King এবং Queen কতটা কাছাকাছি, বা King - Man + Woman ≈ Queen — এই ধরনের relation one-hot-এ ধরাই যায় না।

**Embedding** এই সমস্যার সমাধান করে।  
Embedding-এ প্রতিটি word-কে একটি dense vector দেওয়া হয়, যেমন:

- King   → [0.2, 0.8, 0.1, ..., 0.4]
- Queen  → [0.3, 0.7, 0.2, ..., 0.5]
- Man    → [0.1, 0.6, 0.0, ..., 0.3]


এখন vector space-এ:
- King এবং Queen কাছাকাছি থাকে
- Man এবং Woman কাছাকাছি থাকে
- এবং বিখ্যাত analogy কাজ করে:  
  **King - Man + Woman ≈ Queen**

এটাই Embedding-এর শক্তি।

## 2. Embedding কেন দরকার? (Why do we need Embeddings?)

Neural Network এবং Deep Learning model-গুলো **সংখ্যা** নিয়ে কাজ করে। তারা সরাসরি text বুঝতে পারে না।

### সমস্যাগুলো:

| সমস্যা | ব্যাখ্যা |
|------|--------|
| **Discrete nature** | Words হলো discrete symbols। Model এগুলোকে সরাসরি process করতে পারে না। |
| **No semantic information** | One-hot encoding-এ কোনো meaning বা similarity নেই। |
| **Curse of Dimensionality** | Vocabulary size হাজার হাজার হলে one-hot vector অনেক বড় এবং sparse হয় → computation expensive + generalization খারাপ হয়। |
| **No generalization** | Model দেখেনি এমন word-এর জন্য কিছুই শিখতে পারে না। |

### Embedding কীভাবে সাহায্য করে:

1. **Dense Representation** → কম dimension-এ বেশি information
2. **Semantic Similarity** capture করে
3. **Transfer Learning** সম্ভব হয় (pre-trained embedding ব্যবহার করা যায়)
4. Model-কে **generalization** করতে সাহায্য করে

## 3. Word Embedding কী?

যখন আমরা **words**-কে vector-এ রূপান্তর করি, তখন তাকে **Word Embedding** বলা হয়।

Word Embedding-এর লক্ষ্য:
> এমন vector শেখা যাতে semantically similar words vector space-এ কাছাকাছি থাকে এবং linguistic relationships (analogy, gender, tense ইত্যাদি) গাণিতিকভাবে capture হয়।

---

## 4. Word2Vec কী?

**Word2Vec** হলো Word Embedding শেখার সবচেয়ে বিখ্যাত এবং প্রভাবশালী algorithm।

এটি ২০১৩ সালে Google-এর Tomas Mikolov এবং তার team তৈরি করেছিলেন।

### মূল Idea (Intuition):

> **"You shall know a word by the company it keeps."**  
> — J.R. Firth (1957)

মানে: একটি শব্দের অর্থ বোঝা যায় সে যেসব শব্দের সাথে থাকে, সেগুলো দেখে।

Word2Vec এই principle-কে কাজে লাগিয়ে দুটি neural network architecture ব্যবহার করে word embedding শেখে:

1. **CBOW** (Continuous Bag of Words)
2. **Skip-gram**

### Word2Vec কী করে?

- বিশাল পরিমাণ text (corpus) দেখে
- প্রতিটি word-এর জন্য একটি dense vector শেখে
- যেসব words একই context-এ আসে, তাদের vector কাছাকাছি হয়
- ফলে semantic এবং syntactic relationship automatically capture হয়

### কেন Word2Vec এত গুরুত্বপূর্ণ?

| কারণ | ব্যাখ্যা |
|------|--------|
| **Efficiency** | আগের method-গুলোর তুলনায় অনেক দ্রুত এবং scalable |
| **Quality** | খুব ভালো quality-এর embedding তৈরি করে |
| **Analogy Capability** | `King - Man + Woman ≈ Queen` এরকম relation ধরে |
| **Foundation** | পরবর্তী অনেক advanced model (GloVe, FastText, BERT ইত্যাদি)-এর ভিত্তি তৈরি করেছে |

Text (Words)

Embedding Layer / Word Embedding

↓

Dense Vector (ভালো)

↓
